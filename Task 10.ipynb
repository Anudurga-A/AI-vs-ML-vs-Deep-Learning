# AI vs ML vs Deep Learning

## Introduction
Artificial Intelligence (AI), Machine Learning (ML), and Deep Learning (DL) are important concepts in computer science. They are used to make machines perform intelligent tasks, learn from data, and recognize complex patterns.

## 1. Artificial Intelligence (AI)
Artificial Intelligence is a technology that enables machines to perform tasks that normally require human intelligence.

**Example:** Chatbots, virtual assistants, and robotics.

## 2. Machine Learning (ML)
Machine Learning is a subset of AI that enables computers to learn from data and make predictions without being explicitly programmed for every task.

**Example:** Email spam detection and movie recommendations.

## 3. Deep Learning (DL)
Deep Learning is a subset of Machine Learning that uses multi-layered neural networks to learn complex patterns from large amounts of data.

**Example:** Face recognition, speech recognition, and image classification.

## 4. Comparison Table

| Feature | AI | ML | Deep Learning |
|---|---|---|---|
| Full Form | Artificial Intelligence | Machine Learning | Deep Learning |
| Meaning | Machines perform intelligent tasks | Machines learn from data | Neural networks learn complex patterns |
| Data Requirement | Depends on the method | Usually requires data | Often requires large amounts of data |
| Example | Rule-based chatbot | Spam detection | Face recognition |
| Applications | Robotics, virtual assistants | Recommendations, prediction | Image and speech recognition |

## 5. Real-World Examples

### AI
AI is used in virtual assistants, chatbots, and robotics to perform intelligent tasks.

### Machine Learning
ML is used by shopping and streaming platforms to recommend products, movies, and videos based on user preferences.

### Deep Learning
Deep Learning is used in face recognition, voice assistants, medical image analysis, and self-driving technology.

## 6. Relationship Between AI, ML, and Deep Learning

AI is the broader field of creating systems that perform intelligent tasks. Machine Learning is a subset of AI that learns from data. Deep Learning is a subset of Machine Learning that uses multi-layered neural networks.

**Relationship:**
Deep Learning is part of Machine Learning, and Machine Learning is part of Artificial Intelligence.

## 7. Conclusion

AI helps machines perform intelligent tasks. ML enables machines to learn from data, while Deep Learning helps them identify complex patterns using neural networks. These technologies are widely used in modern applications such as recommendations, chatbots, and image recognition.
